1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter.

```python
class counter:
  def __init__(self, max):
    self.value = 0
    self.max = max
  def increment(self):
    if self.value >= self.max:
      print("Error: maximum value reached")
    else: self.value += 1
  def reset(self):
    self.value = 0
  def __repr__(self):
    return f"Counter(value={self.value}, max_value={self.max})"
```

```python
import math



class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (x2 - x1) / (y2 - y1)
        for y in range(y1, y2):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))


class shape:
    def __init__(self, x, y, name=""):
        self._x = x
        self._y = y
        self.name = name

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        pass

    def perimeter(self):
        pass

    def perimeter_points(self, n=16):
        pass

    def point_inside(self, px, py):
        pass

    def overlaps(self, other, samples=32):
        for (px, py) in self.perimeter_points(samples):
            if other.point_inside(px, py):
                return True
        for (px, py) in other.perimeter_points(samples):
            if self.point_inside(px, py):
                return True
        return False

    def paint(self, canvas, char='*'):
        pass



class rectangle(shape):
    def __init__(self, x, y, length, width, name=""):
        shape.__init__(self, x, y, name)
        self.__length = length
        self.__width = width

    def get_length(self): return self.__length
    def get_width(self): return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self, n=16):
        x, y, l, w = self._x, self._y, self.__length, self.__width
        corners = [(x, y), (x + l, y), (x + l, y + w), (x, y + w)]
        edges = [(corners[i], corners[i - 1]) for i in range(4)]
        perim = self.perimeter()
        pts = []
        for i in range(n):
            d = perim * i / n
            for a, b in edges:
                elen = math.dist(a, b)
                if d <= elen:
                    t = d / elen if elen else 0
                    pts.append((a[0] + t * (b[0] - a[0]), a[1] + t * (b[1] - a[1])))
                    break
                d -= elen
        return pts

    def point_inside(self, px, py):
        x, y, l, w = self._x, self._y, self.__length, self.__width
        xmin, xmax = min(x, x + l), max(x, x + l)
        ymin, ymax = min(y, y + w), max(y, y + w)
        return xmin <= px <= xmax and ymin <= py <= ymax

    def paint(self, canvas, char='*'):
        x, y, l, w = round(self._x), round(self._y), round(self.__length), round(self.__width)
        canvas.h_line(x, y, w + 1, char=char)
        canvas.h_line(x + l, y, w + 1, char=char)
        canvas.v_line(x, y, l + 1, char=char)
        canvas.v_line(x, y + w, l + 1, char=char)



class circle(shape):
    def __init__(self, x, y, radius, name=""):
        shape.__init__(self, x, y, name)
        self.__radius = radius

    def get_radius(self): return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self, n=16):
        r = self.__radius
        return [(self._x + r * math.cos(2 * math.pi * i / n),
                 self._y + r * math.sin(2 * math.pi * i / n)) for i in range(n)]

    def point_inside(self, px, py):
        return (px - self._x) ** 2 + (py - self._y) ** 2 <= self.__radius ** 2

    def paint(self, canvas, char='*'):
        n = max(16, round(self.perimeter() * 2))
        for (px, py) in self.perimeter_points(n):
            row, col = round(px), round(py)
            if 0 <= row < canvas.height and 0 <= col < canvas.width:
                canvas.set_pixel(row, col, char)



class triangle(shape):
    def __init__(self, x1, y1, x2, y2, x3, y3, name=""):
        shape.__init__(self, x1, y1, name)
        self.__x2, self.__y2 = x2, y2
        self.__x3, self.__y3 = x3, y3

    def get_vertices(self):
        return (self._x, self._y), (self.__x2, self.__y2), (self.__x3, self.__y3)

    def area(self):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        return abs((x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)) / 2)

    def perimeter(self):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        dist = lambda ax, ay, bx, by: math.hypot(bx - ax, by - ay)
        return dist(x1, y1, x2, y2) + dist(x2, y2, x3, y3) + dist(x3, y3, x1, y1)

    def perimeter_points(self, n=16):
        verts = list(self.get_vertices())
        edges = [(verts[i], verts[i - 1]) for i in range(len(verts))]
        perim = self.perimeter()
        pts = []
        for i in range(n):
            d = perim * i / n
            for a, b in edges:
                elen = math.dist(a, b)
                if d <= elen:
                    t = d / elen if elen else 0
                    pts.append((a[0] + t * (b[0] - a[0]), a[1] + t * (b[1] - a[1])))
                    break
                d -= elen
        return pts

    def point_inside(self, px, py):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        def sign(ax, ay, bx, by, cx, cy):
            return (ax - cx) * (by - cy) - (bx - cx) * (ay - cy)
        d1 = sign(px, py, x1, y1, x2, y2)
        d2 = sign(px, py, x2, y2, x3, y3)
        d3 = sign(px, py, x3, y3, x1, y1)
        has_neg = d1 < 0 or d2 < 0 or d3 < 0
        has_pos = d1 > 0 or d2 > 0 or d3 > 0
        return not (has_neg and has_pos)

    def paint(self, canvas, char='*'):
        n = max(16, round(self.perimeter() * 2))
        for (px, py) in self.perimeter_points(n):
            row, col = round(px), round(py)
            if 0 <= row < canvas.height and 0 <= col < canvas.width:
                canvas.set_pixel(row, col, char)



class CompoundShape(shape):
    def __init__(self, shapes, name=""):
        x0, y0 = (shapes[0].get_x(), shapes[0].get_y()) if shapes else (0, 0)
        shape.__init__(self, x0, y0, name)
        self.shapes = list(shapes)

    def area(self):
        return sum(s.area() for s in self.shapes)

    def perimeter(self):
        return sum(s.perimeter() for s in self.shapes)

    def point_inside(self, px, py):
        return any(s.point_inside(px, py) for s in self.shapes)

    def paint(self, canvas, char='*'):
        for s in self.shapes:
            s.paint(canvas, char=char)
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again.

```python
class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = list()

    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()

        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)

    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)

    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"
        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i += 1
        return name

      # Had to use claude to change code from number 10 as I was getting errors because I did not include the name variable in the classes
```

For example:

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b

    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"

    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()


def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```

```python
import math


class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (x2 - x1) / (y2 - y1)
        for y in range(y1, y2):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))


class shape:
    def __init__(self, x, y, name=""):
        self._x = x
        self._y = y
        self.name = name

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        pass

    def perimeter(self):
        pass

    def perimeter_points(self, n=16):
        pass

    def point_inside(self, px, py):
        pass

    def overlaps(self, other, samples=32):
        for (px, py) in self.perimeter_points(samples):
            if other.point_inside(px, py):
                return True
        for (px, py) in other.perimeter_points(samples):
            if self.point_inside(px, py):
                return True
        return False

    def paint(self, canvas, char='*'):
        pass


class rectangle(shape):
    def __init__(self, x, y, length, width, name=""):
        shape.__init__(self, x, y, name)
        self.__length = length
        self.__width = width

    def get_length(self): return self.__length
    def get_width(self): return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self, n=16):
        x, y, l, w = self._x, self._y, self.__length, self.__width
        corners = [(x, y), (x + l, y), (x + l, y + w), (x, y + w)]
        edges = [(corners[i], corners[i - 1]) for i in range(4)]
        perim = self.perimeter()
        pts = []
        for i in range(n):
            d = perim * i / n
            for a, b in edges:
                elen = math.dist(a, b)
                if d <= elen:
                    t = d / elen if elen else 0
                    pts.append((a[0] + t * (b[0] - a[0]), a[1] + t * (b[1] - a[1])))
                    break
                d -= elen
        return pts

    def point_inside(self, px, py):
        x, y, l, w = self._x, self._y, self.__length, self.__width
        xmin, xmax = min(x, x + l), max(x, x + l)
        ymin, ymax = min(y, y + w), max(y, y + w)
        return xmin <= px <= xmax and ymin <= py <= ymax

    def paint(self, canvas, char='*'):
        x, y, l, w = round(self._x), round(self._y), round(self.__length), round(self.__width)
        canvas.h_line(x, y, w + 1, char=char)
        canvas.h_line(x + l, y, w + 1, char=char)
        canvas.v_line(x, y, l + 1, char=char)
        canvas.v_line(x, y + w, l + 1, char=char)

    def __repr__(self):
        return f"rectangle({self._x},{self._y},{self.__length},{self.__width},{self.name!r})"


class circle(shape):
    def __init__(self, x, y, radius, name=""):
        shape.__init__(self, x, y, name)
        self.__radius = radius

    def get_radius(self): return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self, n=16):
        r = self.__radius
        return [(self._x + r * math.cos(2 * math.pi * i / n),
                 self._y + r * math.sin(2 * math.pi * i / n)) for i in range(n)]

    def point_inside(self, px, py):
        return (px - self._x) ** 2 + (py - self._y) ** 2 <= self.__radius ** 2

    def paint(self, canvas, char='*'):
        n = max(16, round(self.perimeter() * 2))
        for (px, py) in self.perimeter_points(n):
            row, col = round(px), round(py)
            if 0 <= row < canvas.height and 0 <= col < canvas.width:
                canvas.set_pixel(row, col, char)

    def __repr__(self):
        return f"circle({self._x},{self._y},{self.__radius},{self.name!r})"


class triangle(shape):
    def __init__(self, x1, y1, x2, y2, x3, y3, name=""):
        shape.__init__(self, x1, y1, name)
        self.__x2, self.__y2 = x2, y2
        self.__x3, self.__y3 = x3, y3

    def get_vertices(self):
        return (self._x, self._y), (self.__x2, self.__y2), (self.__x3, self.__y3)

    def area(self):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        return abs((x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)) / 2)

    def perimeter(self):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        dist = lambda ax, ay, bx, by: math.hypot(bx - ax, by - ay)
        return dist(x1, y1, x2, y2) + dist(x2, y2, x3, y3) + dist(x3, y3, x1, y1)

    def perimeter_points(self, n=16):
        verts = list(self.get_vertices())
        edges = [(verts[i], verts[i - 1]) for i in range(len(verts))]
        perim = self.perimeter()
        pts = []
        for i in range(n):
            d = perim * i / n
            for a, b in edges:
                elen = math.dist(a, b)
                if d <= elen:
                    t = d / elen if elen else 0
                    pts.append((a[0] + t * (b[0] - a[0]), a[1] + t * (b[1] - a[1])))
                    break
                d -= elen
        return pts

    def point_inside(self, px, py):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        def sign(ax, ay, bx, by, cx, cy):
            return (ax - cx) * (by - cy) - (bx - cx) * (ay - cy)
        d1 = sign(px, py, x1, y1, x2, y2)
        d2 = sign(px, py, x2, y2, x3, y3)
        d3 = sign(px, py, x3, y3, x1, y1)
        has_neg = d1 < 0 or d2 < 0 or d3 < 0
        has_pos = d1 > 0 or d2 > 0 or d3 > 0
        return not (has_neg and has_pos)

    def paint(self, canvas, char='*'):
        n = max(16, round(self.perimeter() * 2))
        for (px, py) in self.perimeter_points(n):
            row, col = round(px), round(py)
            if 0 <= row < canvas.height and 0 <= col < canvas.width:
                canvas.set_pixel(row, col, char)

    def __repr__(self):
        (x1, y1), (x2, y2), (x3, y3) = self.get_vertices()
        return f"triangle({x1},{y1},{x2},{y2},{x3},{y3},{self.name!r})"


class CompoundShape(shape):
    def __init__(self, shapes, name=""):
        x0, y0 = (shapes[0].get_x(), shapes[0].get_y()) if shapes else (0, 0)
        shape.__init__(self, x0, y0, name)
        self.shapes = list(shapes)

    def area(self):
        return sum(s.area() for s in self.shapes)

    def perimeter(self):
        return sum(s.perimeter() for s in self.shapes)

    def point_inside(self, px, py):
        return any(s.point_inside(px, py) for s in self.shapes)

    def paint(self, canvas, char='*'):
        for s in self.shapes:
            s.paint(canvas, char=char)

    def __repr__(self):
        return f"CompoundShape([{','.join(repr(s) for s in self.shapes)}],{self.name!r})"
```

```python
class RasterDrawing:
    def __init__(self, shapes=None):
        self.shapes = dict()
        self.shape_names = list()
        for s in (shapes or []):
            self.add_shape(s)

    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()
        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)

    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)

    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"
        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i += 1
        return name

    def __repr__(self):
        shape_reprs = ",".join(repr(self.shapes[name]) for name in self.shape_names)
        return f"RasterDrawing([{shape_reprs}])"

    def save(self, filename):
        f = open(filename, "w")
        f.write(self.__repr__())
        f.close()


def raster_drawing_loader(filename):
    f = open(filename, "r")
    tmp = eval(f.read())
    f.close()
    return tmp
```

```python
drawing = RasterDrawing()
drawing.add_shape(rectangle(2, 2, 10, 20, name="box"))
drawing.add_shape(circle(20, 40, 8))
drawing.add_shape(triangle(30, 5, 38, 5, 34, 20))

canvas = Canvas(60, 40)
drawing.paint(canvas)
canvas.display()
drawing.save("MyDrawing.raster")
```
